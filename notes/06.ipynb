## 2. Operacje na słownikach

```python
dic = {"hi": "cześć", "bye": "papa"}
len(dic)
```

```python
dic["bye"]       # Wyciąga element ze słownika po kluczu
```

```python
dic["python"]    # Odwołanie się do klucza, który nie istnieje zwraca błąd
```

```python
dic = {"hi": "cześć", "bye": "papa"}
"hi" in dic                            # Sprawdzenie, czy klucz istnieje w słowniku
```

```python
dic = {"hi": "cześć", "bye": "papa"}
"hi" not in dic                       # Sprawdzenie, czy klucz nie istnieje w słowniku
```

Operator `in` używa różnych algorytmów dla list i słowników. W przypadku list, używa algorytmu wyszukiwania liniowego. Gdy lista staje się dłuższa, czas wyszukiwania wydłuża się wprost proporcjonalnie do jej długości. W przeciwieństwie do list, które muszą sprawdzać każdy element po kolei, słowniki zapamiętują położenie każdego klucza dzięki specjalnej funkcji matematycznej zwanej hash, co sprawia, że wyszukiwanie jest bardzo szybkie, zajmując mniej więcej tyle samo czasu, niezależnie od tego, ile elementów jest w słowniku.

```python
dic = {"hi": "cześć", "bye": "papa"}
if "hi" in dic:
    value = dic["hi"]
    print(value)
else:
    print("Sorry, no translation")
```

```python
# Zlicza występowanie każdego znaku w tekście użytkownika (wersja 1)
text = input("Enter a text: ")
counts = {}
for letter in text:
    if letter not in counts:
        counts[letter] = 1
    else:
        counts[letter] += 1
print(counts)
```

```python
# Zlicza występowanie każdego znaku w tekście użytkownika (wersja 1)
text = input("Enter a text: ")
counts = {}
for letter in text:
    if letter not in counts:
        counts[letter] = 0
    counts[letter] += 1
print(counts)
```

## 4. Metody słowników

#### Metoda `update()`
Łączy klucze i wartości danego słownika z innym słownikiem

```python
d1 = {1: "a", 2: "b"}
d2 = {2: "c", 3: "d"}

d1.update(d2)
print(d1)
```

#### Metoda `pop()`
Usuwa element ze słownika w oparciu o podanay klucz i zwraca jego wartość

```python
dic = {"hi": "cześć", "bye": "papa"}
dic.pop("hi")
print(dic)
```

```python
dic = {"hi": "cześć", "bye": "papa"}
deleted_element = dic.pop("hi")
print(deleted_element)
```

##### Instrukcja `del`

```python
dic = {"hi": "cześć", "bye": "papa"}
del dic["hi"]
print(dic)
```

#### Metoda `get`

Metoda `get` wyciąga element po kluczu, a jeśli nie taki klucz nie istnieje to zwraca Nona. Dzięki tej metodzie nie musimy ręcznie sprawdzać, czy klucz istnieje w słowniku przed jego wyciągnięciem.

```python
# Ręczne sprawdzenie, czy klucz istnieje w słowniku zanim go wyciągniemy.
dic = {"hi": "cześć", "bye": "papa"}
if "hi" in dic:
    translation = dic["hi"]
else:
    translation = None
```

```python
# Korzystając z metody get możemy zrobić to samo co wyżej, ale krócej.
dic = {"hi": "cześć", "bye": "papa"}
translation = dic.get("hi")
print(translation)
```

```python
dic = {"hi": "cześć", "bye": "papa"}
translation = dic.get("ten")
print(translation)
```

```python
# Metoda get pozwala nam ustawić wartość dla nieistniejącego klucza.
dic = {"hi": "cześć", "bye": "papa"}
translation = dic.get("ten", "Sorry no translation")
print(translation)
```

```python
# Zlicza występowanie każdego znaku w tekście użytkownika (wersja 3)
text = "abcdgjshjdhsj"
counts = {}
for letter in text:
    counts[letter] = counts.get(letter, 0) + 1
print(counts)
```

#### Metody `keys()`, `values()`, `items()`

```python
dic = {"hi": "cześć", "bye": "papa"}
dic.keys()                             # Zwraca wszystkie klucze
```

```python
dic = {"hi": "cześć", "bye": "papa"}
dic.values()                           # Zwraca wszystkie wartości
```

```python
# Zwraca klucze i odpowiadające wartości jako lista
# krotek dwuelementowych, gdzie pierwszy element krotki to klucz
# a drugi element krotki do wartość
dic.items()
```

## 5. Iterowanie po słowniku

```python
# 1. Klasyczny sposób
# Pętla for na słowniku iteruje po kluczach słownika.
dic = {"hi": "cześć", "bye": "papa"}
for key in dic:
    print(key, dic[key])
```

```python
# 2. Za pomocą metody items (często używany sposób)
dic = {"hi": "cześć", "bye": "papa"}
for key, value in dic.items():
    print(key, value)
```

## 6. Inne sposoby tworzenia słowników

```python
# Funkcja dict oprócz tego, że tworzy pusty słownik, może przyjąć
# listę krotek dwuelementowych.
lst = [("Python", 1), ("C++", 2), ("JavaScript", 3)]
dic = dict(lst)
print(dic)
```

```python
# Ręczne zbudowanie słownika z listy krotek dwuelementowych
lst = [("Python", 1), ("C++", 2), ("JavaScript", 3)]
dic = {}
for key, value in lst:
    dic[key] = value
print(dic)
```
