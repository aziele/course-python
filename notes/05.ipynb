## 1.2. Metody list

#### Metoda `append()`

```python
courses = ["Python", "Linux"]

courses.append("Biology")      # Dodaje pojedynczy element na koniec listy
print(courses)
```

```python
courses = ["Python", "Linux"]

courses = courses.append("Biology")
print(courses)
```

#### Metoda `extend()`

```python
lst = ["a", "b", "c"]

lst.extend(["d", "e", "f"])

print(lst)
```

```python
lst = ["a", "b", "c"]

lst.append(["d", "e", "f"])

print(lst)
```

```python
lst = ["a", "b", "c"]

lst.append("d", "e", "f")   # append tylko pojedynczy obiekt

print(lst)
```

#### Metoda `sort()`

```python
lst = ["d", "b", "a", "c"]

lst.sort()
print(lst)
```

```python
lst = ["d", "b", "a", "c"]

lst.sort(reverse=True)
print(lst)
```

```python
lst = ["d", "b", "a", "c"]

lst = lst.sort(reverse=True)
print(lst)
```

#### Metoda `reverse()`

```python
lst = ["Python", "Math", "Statistics"]

lst.reverse()

print(lst)
```

#### Metoda `pop()`

```python
lst = ["Python", "Math", "Statistics"]

lst.pop()

print(lst)
```

```python
lst = ["Python", "Math", "Statistics"]

deleted_element = lst.pop()
print(deleted_element)
```

```python
lst = ["Python", "Math", "Statistics"]

lst.pop()

print(lst)
```

```python
lst = ["Python", "Math", "Statistics"]

lst.pop(1)               # Usuwa po pozycji

print(lst)
```

#### Metoda `remove()`

```python
lst = ["Python", "Math", "Statistics"]

lst.remove("Math")        # Usuwa po wartości

print(lst)
```

```python
lst = ["Python", "Math", "Statistics", "Math", "Math"]

lst.remove("Math")        # Usuwa po wartości

print(lst)
```

#### Metoda `insert()`

```python
lst = ["Python", "Math", "Statistics"]

lst.insert(1, "Biology")

print(lst)
```

```python
lst = ["Python", "Math", "Statistics"]

lst.insert(1, [1, 2, 3])

print(lst)
```

#### Metoda `count()`

```python
lst = ["Python", "Math", "Statistics", "Math"]

lst.count("Math")
```

#### Metoda `index()`

```python
lst = ["Python", "Math", "Statistics", "Math"]

lst.index("Math")
```

```python
lst = ["Python", "Math", "Statistics", "Math"]

lst.index("Biology")      # Błąd
```

## 1.3. Funkcje list

```python
lst = [3, 2, 4, 1]

lst.sort()

print(lst)
```

```python
lst = [3, 2, 4, 1]

new_lst = sorted(lst)

print(new_lst)
```

```python
lst = [3, 2, 4, 1]

new_lst = sorted("dcba")

print(new_lst)
```

```python
lst = [3, 2, 4, 1]

new_lst = sorted("dcba", reverse=True)

print(new_lst)
```

```python
numbers = [1, 2, 5, 3.14, 9.0]


print(len(numbers))
print(min(numbers))
print(max(numbers))
print(sum(numbers))
```

```python
numbers = [1, 2, 5, 3.14, 9.0]

total_sum = 0
for number in numbers:
    total_sum += number

print(total_sum)
```

## 1.4. Listy i stringi

```python
csv = "Gene1,+,1000,50000,description"

lst = csv.split(",")
print(lst)
```

```python
csv = "Gene1,+,1000,50000,description"

lst = csv.split(",")
lst[1] = "-"
csv = ",".join(lst)
print(csv)
```
