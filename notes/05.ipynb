## 1.2. Metody list

#### Metoda `append()`

```python
courses = ["Python", "Linux"]
courses.append("Biology")      # Dodaje pojedynczy element na koniec listy
print(courses)
```

```python
courses = ["Python", "Linux"]
courses = courses.append("Biology") # To nie działa, bo metoda append modyfikuje listę w miejscu (nie zwraca nowej)
print(courses)
```

#### Metoda `extend()`

```python
# Dodaje wiele elementów na koniec listy.
# Innymi słowy, rozszerza jedną listą dodając do niej elementy drugiej listy.

lst = ["a", "b", "c"]
lst.extend(["d", "e", "f"])
print(lst)
```

```python
lst = ["a", "b", "c"]
lst.append(["d", "e", "f"])   # Metoda append umieszcza jeden element.
print(lst)
```

#### Metoda `sort()`

```python
lst = ["d", "b", "a", "c"]
lst.sort()
print(lst)
```

```python
lst = ["d", "b", "a", "c"]
lst.sort(reverse=True)        # Malejąco
print(lst)
```

```python
lst = ["d", "b", "a", "c"]

lst = lst.sort(reverse=True)  # To nie działa, bo sort modyfikuje listę w miejscu (nie tworzy nowej)
print(lst)
```

#### Metoda `reverse()`

```python
lst = ["Python", "Math", "Statistics"]
lst.reverse()
print(lst)
```

#### Metoda `pop()`

```python
lst = ["Python", "Math", "Statistics"]
lst.pop()        # Usuwa ostatni element
print(lst)
```

```python
lst = ["Python", "Math", "Statistics"]

deleted_element = lst.pop() # Zwraca również usunięty element.
print(deleted_element)
```

```python
lst = ["Python", "Math", "Statistics"]
lst.pop()
print(lst)
```

```python
lst = ["Python", "Math", "Statistics"]
lst.pop(1)               # Usuwa element po pozycji
print(lst)
```

#### Metoda `remove()`

```python
lst = ["Python", "Math", "Statistics"]
lst.remove("Math")        # Usuwa element po wartości
print(lst)
```

```python
lst = ["Python", "Math", "Statistics", "Math", "Math"]
lst.remove("Math")        # Usuwa pierwsze wystąpienie elementu
print(lst)
```

#### Metoda `insert()`

```python
lst = ["Python", "Math", "Statistics"]
lst.insert(1, "Biology")     # Umieszcza pojedynczy element w danej pozycji w liście.
print(lst)
```

```python
lst = ["Python", "Math", "Statistics"]
lst.insert(1, [1, 2, 3])
print(lst)
```

#### Metoda `count()`

```python
lst = ["Python", "Math", "Statistics", "Math"]
lst.count("Math")  # Zlicza wystąpienia elementu
```

#### Metoda `index()`

```python
lst = ["Python", "Math", "Statistics", "Math"]
lst.index("Math")       # Zwraca indeks na którym znajduje się element
```

```python
lst = ["Python", "Math", "Statistics", "Math"]
lst.index("Biology")      # Błąd jeśli elementu nie ma w liście
```

## 1.3. Funkcje list

```python
lst = [3, 2, 4, 1]
lst.sort()         # To jest metoda, która modyfikuje listę w miejscu
print(lst)
```

```python
lst = [3, 2, 4, 1]
new_lst = sorted(lst) # Funkcja sorted zwraca nową posortowaną listę, a listę oryginalną zachowuje.
print(new_lst)
```

```python
lst = [3, 2, 4, 1]
new_lst = sorted("dcba")  # Funkcją sorted możemy sortować stringi, ale zwracana jest zawsze lista.
print(new_lst)
```

```python
lst = [3, 2, 4, 1]
new_lst = sorted("dcba", reverse=True)   # W kolejności malejącej.
print(new_lst)
```

```python
numbers = [1, 2, 5, 3.14, 9.0]
print(len(numbers))   # Długość listy
print(min(numbers))   # Najmniejsza wartość w liście (może też być string)
print(max(numbers))   # Największa wartość w liscie (może też być string)
print(sum(numbers))   # Suma liczb (tylko dla liczb!)
```

```python
# Ręczne obliczenie sumy liczb w liście
numbers = [1, 2, 5, 3.14, 9.0]
total_sum = 0
for number in numbers:
    total_sum += number
print(total_sum)
```

## 1.4. Listy i stringi

```python
name = "Python"
lst = list(name)    # Zamienia na listę, ale każda litera jest osobnym elementem
print(lst)
```

```python
name = "Hello! Have a nice day"
lst = list(name)
print(lst)
```

```python
name = "Hello! Have a nice day"
lst = name.split()      # Granicą elementów jest biały znak
print(lst)
```

```python
csv = "Gene1,+,1000,50000,description"
lst = csv.split(",")   # Granicą elementów jest przecinek
print(lst)
```

```python
lst = ['Gene1', '+', '1000', '50000', 'description']
",".join(lst)         # Zamienia listę na string.
```

```python
# Jak zrobić mutację w określonej pozycji w sekwencji.
# Trzeba zamienić na listę (bo stringi są niemutowalne)
# Wprowadzić zmianę w liście
# I zamienić listę z powrotem na string.

seq = "TGCAT"
seq_lst = list(seq)
seq_lst[1] = "C"
seq = "".join(seq_lst)
print(seq)
```

```python
csv = "Gene1,+,1000,50000,description"
lst = csv.split(",")
lst[1] = "-"
csv = ",".join(lst)
print(csv)
```

# 2. Krotki (ang. *tuples*)

Krotka, tak jak lista, jest uporzadkowaną kolekcją obiektów - zachowuje kolejność elementów od lewej do prawej strony. Działają dokładnie tak jak listy, jednak nie mogą być modyfikowane w miejscu (są niemutowalne) i są zapisywane jako seria elementów w zwykłych nawiasach okrągłych, a nie w nawiasach kwadratowych.

```python
tup = (1, 2, 3)
```

```python
# Pusta krotka
tup = ()
tup = tuple()
```

```python
tup = (1, 2, 3)
tup[1]
```

```python
tup = (1, 2, 3)
list(tup)
```

```python
tuple([1, 2, 3])
```

O użyteczności krotek będziemy mówić więcej przy okazji słowników.
