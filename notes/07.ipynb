## 1. Otwieranie plików

```python
open("myfile.txt", "r")     # r - read (tryb wczytywania pliku)
open("outfile.txt", "w")    # w - write (tryb zapisywanie do pliku)
open("infile.txt", "a")     # a - append (zapis na koniec istniejącego pliku)
open("infile.txt", "x")     # plik do zapisu, ale jeżeli plik istnieje to zwraca błąd
```

```python
file = open("myfile.txt")   # tworzy obiekt pliku, który jest łączem do rzeczywistego pliku na dysku
print(file.name)
print(file.mode)
```

## 2. Zamykanie plików

```python
# Klasyczna składnia, ale trzeba pamiętać o zamknięciu pliku

file = open("myfile.txt")
# Tutaj będziemy pracować z plikiem
file.close()
```

```python
# Preferowany sposób (najbardziej bezpieczny), bo mamy gwarancję, że plik zostanie zamknięty
with open("myfile.txt") as file:
    # Tutaj jest praca z plikiem
    print(file.name)
```

## 3. Wczytywanie plików

### 3.1. Pętla for

```python
with open("myfile.txt") as file:
    for line in file:
        print(line)                # W pamięci jest tylko jeden wiersz (nie cały plik)
```

```python
with open("myfile.txt") as file:
    for line in file:
        print(line, end="")
```

```python
file = open("myfile.txt")
for line in file:
    print(line, end="")
file.close()
```

### 3.2. Metoda `readlines()`

```python
with open("myfile.txt") as file:
    lines = file.readlines()         # Wczytanie całego pliku do pamięci jako lista wierszy
    print(lines)
```

```python
with open("myfile.txt") as file:
    lines = []
    for line in file:
        lines.append(line)
    print(lines)
```

### 3.3. Metoda `readline()`

```python
with open("myfile.txt") as file:
    line = file.readline()          # Wczytanie pierwszego wiersza.
    print(line)

    line = file.readline()          # Wczytanie drugiego wiersza.
    print(line)

    line = file.readline()          # Wczytanie trzeciego wiersza.
    print(line)
```

### 3.4. Metoda `read()`

```python
with open("myfile.txt") as file:
    content = file.read()           # Wczytanie całego pliku do pamięci jako string
    print(content)
```

```python
with open("myfile.txt") as file:
    content = file.read(200)         # Wczytanie pierwszych 200 znaków pliku jako string
    print(content)

    print()

    content = file.read(200)        # Wczytanie kolejnych 200 znaków pliku jako string
    print(content)

    content = file.read(200)
    print(content)
```

### 3.5. Ponowne odczytywanie pliku

```python
with open("myfile.txt") as file:
    for line in file:
        print(line, end="")

    for line in file:                # Tego nie zobaczymy, bo możemy tylko raz przejść przez plik.
        print(line)
```

## 4. Zapisywanie plików

### 4.1. Metoda `write()`

```python
with open("outfile.txt", "w") as file:
    file.write("text")

!cat outfile.txt
```

```python
with open("outfile.txt", "w") as file:
    file.write("text")
    file.write("abc")

!cat outfile.txt
```

```python
with open("outfile.txt", "w") as file:
    file.write("text\n")
    file.write("abc")

!cat outfile.txt
```

```python
with open("outfile.txt", "w") as file:
    saved_characters = file.write("text")
    print(saved_characters)
```

### 4.2. Metoda `writelines()`

```python
lines = ["Line 1\n", "Line 2\n", "Line 3\n"]

with open("outfile.txt", "w") as file:
    file.writelines(lines)

!cat outfile.txt
```

```python
lines = ["Line 1\n", "Line 2\n", "Line 3\n"]

with open("outfile.txt", "w") as file:
    for line in lines:
        file.write(line)

!cat outfile.txt
```

## 5. Wczytywanie i zapisywanie

```python
# Klasycznie (ale musimy pamiętać o zamknięciu plików)
infile = open("myfile.txt")
outfile = open("outfile.txt", "w")

for line in infile:
    outfile.write(line)

infile.close()
outfile.close()
```

```python
# Preferowany sposób
with open("myfile.txt") as infile, open("outfile.txt", "w") as outfile:
    for line in infile:
        outfile.write(line)
```

```python
# Dla czytelności możemy rozdzielić na dwa wiersze (często spotykane)
with (
    open("myfile.txt") as infile,
    open("outfile.txt", "w") as outfile
):
    for line in infile:
        outfile.write(line)
```
